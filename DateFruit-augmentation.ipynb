```python
from PIL import Image, ImageEnhance, ImageFilter
import numpy as np
import matplotlib.pyplot as plt

# Load the image
image = Image.open('Sugaey155.jpg')

# Define augmentations
vertical_flip = image.transpose(Image.FLIP_TOP_BOTTOM)
horizontal_flip = image.transpose(Image.FLIP_LEFT_RIGHT)
both_flip = vertical_flip.transpose(Image.FLIP_LEFT_RIGHT)



# Display augmentations
fig, axes = plt.subplots(1, 4, figsize=(20, 10))
axes[0].imshow(image)
axes[0].set_title("Original")
axes[1].imshow(vertical_flip)
axes[1].set_title("Vertical Flip")
axes[2].imshow(horizontal_flip)
axes[2].set_title("Horizontal Flip")
axes[3].imshow(both_flip)
axes[3].set_title("Both Flips")
plt.show()
```

```python
from PIL import Image
import os

# Set the path to your main dataset folder
dataset_folder = 'Date_Fruit/train'

# Loop through each subfolder in the dataset folder
for subfolder in os.listdir(dataset_folder):
    subfolder_path = os.path.join(dataset_folder, subfolder)
    
    # Check if the current path is a folder
    if os.path.isdir(subfolder_path):
        # Loop through each image in the subfolder
        for filename in os.listdir(subfolder_path):
            if filename.endswith(('.png', '.jpg', '.jpeg')):  # Check for image files
                image_path = os.path.join(subfolder_path, filename)
                image = Image.open(image_path)

                # Apply vertical, horizontal, and both flips
                vertical_flip = image.transpose(Image.FLIP_TOP_BOTTOM)
                horizontal_flip = image.transpose(Image.FLIP_LEFT_RIGHT)
                both_flip = vertical_flip.transpose(Image.FLIP_LEFT_RIGHT)

                # Save the augmented images in the same subfolder with modified filenames
                base_name, ext = os.path.splitext(filename)
                vertical_flip.save(os.path.join(subfolder_path, f"{base_name}_vertical{ext}"))
                horizontal_flip.save(os.path.join(subfolder_path, f"{base_name}_horizontal{ext}"))
                both_flip.save(os.path.join(subfolder_path, f"{base_name}_both{ext}"))

print("Augmentation complete! Augmented images are saved in their respective subfolders.")
```
